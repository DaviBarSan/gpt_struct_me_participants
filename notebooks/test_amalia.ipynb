```python
import openai
import os

client = openai.OpenAI(
    api_key=os.getenv("OPENAI_API_KEY"),
    base_url="http://amalia.inesctec.pt:8000/v1"
)

completion = client.chat.completions.create(
    model="amalia-base",
    messages=[{"role": "user", "content": """Generated prompt: Task:


Extract and classify all participants

Classes:
Person, Organization, Object, Location, Nature, Facility, and Other

Format:
JSON-parseable list where each element is a list with two strings, only with the JSON array and nothing else. Do not include any introductory text, explanation, or markdown c
ode fences.The first string is the entity and the second is the class. Place the JSON parsable list inside <Output> tags.
Use encoding: utf-8.

Input:
"PJ detém homem suspeito de balear jovem em coletividade de Setúbal
A Polícia Judiciária (PJ) de Setúbal deteve um homem de 63 anos suspeito de ter baleado um jovem, na passada quarta-feira, no restaurante de uma coletividade de Setúbal, foi 
hoje anunciado.
Segundo um comunicado da PJ, o presumível agressor terá disparado sobre a vítima na sequência de um conflito com um grupo de convivas que se encontrava a almoçar na coletivid
ade, no Alto da Guerra.
A polícia adianta que o sexagenário acabou por alvejar um jovem do grupo com um disparo de arma de fogo, provocando-lhe ferimentos consideráveis num ombro e braço.
O detido, que fugiu do local, está indiciado pelos crimes de homicídio qualificado na forma tentada, detenção de arma proibida e omissão de auxílio.
Já foi presente a primeiro interrogatório judicial, tendo-lhe sido aplicada a medida de coação de prisão preventiva."""}]
)

print(completion.choices[0].message.content)
```
